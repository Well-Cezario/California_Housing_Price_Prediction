```python
import folium
import geopandas as gpd
import json
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

from folium import plugins
from shapely.geometry import Point

from src.config import DADOS_GEO_ORIGINAIS, DADOS_LIMPOS, DADOS_GEO_MEDIAN
from src.graficos import SCATTER_ALPHA, PALETTE

sns.set_theme(palette="bright", style="white")
```

## Overview
This notebook builds interactive maps to visualize housing value and income patterns in a more dynamic way.

```python
df = pd.read_parquet(DADOS_LIMPOS)

df.head()
```

```python
with open(DADOS_GEO_ORIGINAIS, 'r') as file:
    geojson = json.load(file)
```

```python
gdf_counties = gpd.read_parquet(DADOS_GEO_MEDIAN)

gdf_counties.info()
```

## Creating an Interactive heatmap with median house value per Count

### Main findings
The map confirms that richer and more expensive areas are concentrated in certain coastal and high-demand regions.

```python
size_map = {'width': 700, 'height':500}
map_center = [df['latitude'].mean(), df['longitude'].mean()]
```

```python
fig_house_value = folium.Figure(**size_map)
map_house_value = folium.Map(
    location=map_center,
    zoom_start=5,
    tiles='https://tile.openstreetmap.de/{z}/{x}/{y}.png',
    attr='&copy; <a href="https://www.openstreetmap.org/copyright">OpenStreetMap</a> contributors',
    control_scale=True
).add_to(fig_house_value)

folium.Choropleth(
    geo_data=geojson,
    name='Choropleth_house_value',
    data=gdf_counties,
    columns=['abbrev', 'median_house_value'],
    key_on='feature.properties.abbrev',
    legend_name='MEDIAN HOUSE VALUE',
    fill_color='YlGn',
    fill_opacity=0.5,
    
).add_to(map_house_value)

folium.LayerControl().add_to(map_house_value)
plugins.MousePosition().add_to(map_house_value)
folium.LatLngPopup().add_to(map_house_value)

map_house_value
```

### Cerating an interactive heatmap with Median Income per Count

```python
fig_median_income = folium.Figure(**size_map)

map_median_income = folium.Map(
    location=map_center,
    zoom_start=5,
    tiles='https://tile.openstreetmap.de/{z}/{x}/{y}.png',
    attr='&copy; <a href="https://www.openstreetmap.org/copyright">OpenStreetMap</a> contributors',
    control_scale=True
).add_to(fig_median_income)

folium.Choropleth(
    geo_data=geojson,
    name='Choropleth_median_income',
    data=gdf_counties,
    columns=['abbrev', 'median_income'],
    key_on='feature.properties.abbrev',
    legend_name='MEDIAN_INCOME',
    fill_opacity=0.5,
    
).add_to(map_median_income)

folium.LayerControl().add_to(map_median_income)
plugins.MousePosition().add_to(map_median_income)
folium.LatLngPopup().add_to(map_median_income)

map_median_income
```

## Main findings & Considerations
The maps show that coastal and near-bay areas tend to have higher prices. Inland and more distant areas are usually cheaper.
Interactive maps are useful for storytelling, but the actual model still depends on structured features. Visualization helps guide feature selection and interpretation.
